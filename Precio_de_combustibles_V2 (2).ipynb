## OUTLIERS

```python
def find_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    return outliers
```

```python
# Identificar outliers en la columna 'precio' usando la función
outliers_precio = find_outliers_iqr(data_set_pc, 'precio')

print(f"Se encontraron {len(outliers_precio)} outliers en la columna 'precio'.")
print("Valores atípicos en 'precio' (primeros 5):")
display(outliers_precio.head())
```

# Análisis gráfico

Gráfico de Barras de Precios Promedio por Producto
Este gráfico de barras muestra el precio promedio de cada tipo de combustible, lo que facilita la comparación visual entre ellos.


```python
average_prices = data_set_pc.groupby('producto')['precio'].mean().reset_index()
average_prices = average_prices.sort_values(by='precio', ascending=False)
average_prices['precio'] = average_prices['precio'].round(2)

fig = px.bar(average_prices, x='producto', y='precio',
             title='Precio Promedio por Tipo de Combustible',
             labels={'producto': 'Tipo de Combustible', 'precio': 'Precio Promedio'})
fig.update_layout(xaxis_title_standoff=25) # Ajusta el espaciado del título del eje X
fig.show()
```

Por lo que puedo interpretar, el Gas Oil Grado 3 y la Nafta Premium tienen casi el mismo valor, asi como el Gas Oil Grado 2 y la Nafta Super. El GNC es el producto de menor valor, por lo que explica que los automoviles que se utilizan para trabajr (taxis, aplicaciones, etc) utilicen este combustible para abaratar sus costos.

Histograma Interactivo de Precios por Momento del Día (Plotly)
Este histograma interactivo, generado con Plotly, permite explorar la distribución de los precios de combustible (precio) para cada momento del día (tipohorario).

```python
fig = px.histogram(data_set_pc, y='tipohorario', x='precio',
                   title='Distribución Interactiva de Precios de Combustible por Momento del Día',
                   labels={'precio': 'Precio del Combustible', 'tipohorario': 'Momento del Día de Carga'},
                   histnorm='density', # o 'count' si prefieres la cuenta de observaciones
                   nbins=50, # Número de "contenedores" en el histograma
                   marginal='box', # Añade un boxplot marginal para cada grupo
                   orientation='h') # Orientación horizontal para el histograma
```

```python
fig.update_layout(bargap=0.1) # Espacio entre las barras
fig.show()
```

Por lo que interpreto, no varia el precio del combustible de manera significativa, segun el momento del día

Se puede ver que las marcas que tienen cajas mas grandes tienen mayor variación de precios, mientras que los que tienen cajas mas chicas sus precios varian poco, aunque por ejemplo YPF, posee muchos valores outliers, lo que indica que posee muchos valores por arriba y debajo de los tipicos.
